# [Business Understanding](https://rhindottire.github.io/Data-Mining/dataProcess.html#business-understanding)

## 1. Define the Problem

Collect **200 Indonesian news articles** from detik.com as the dataset for a Web Search & Mining course project, organized following the CRISP-DM methodology. The underlying task of interest is Indonesian news categorization (sport vs finance), which will be addressed in the later phases of this project.

## 2. Set Objectives

- Build a clean, labeled dataset of 200 articles (100 **sport**, 100 **finance**) with the required schema (`id`, `isi_berita`, `label`, `url`).
- Prepare the data so that the next CRISP-DM phases (Data Preparation, Modeling, Evaluation, Production) can be completed in the upcoming assignments.

## 3. Identify Stakeholders

- **Student (me)**: submissions as deliverables of the practical course.
- **Lecturer (MULA'AB, S.Si., M.Kom)**: evaluates the completed work.
- **Readers** of the published Jupyter Book / GitHub Pages.

## 4. Define Success Criteria

- Exactly **200 rows**: 100 `sport` (ids 1-100) and 100 `finance` (ids 101-200).
- **0 missing** values in `isi_berita`; `id` sequential 1-200.
- Dataset loads without errors and matches the required columns (`id`, `isi_berita`, `label`, `url`).

## 5. Assess Situation (Constraints & Ethics)

The collection respects polite, responsible crawling:

- **robots.txt**: detik.com allows crawling (`User-agent: * Allow: /`); a few paths are disallowed for Googlebot only, and the official sitemap is used instead of guessing URLs.
- **Politeness**: every request has a delay of at least **1 second** and a browser-like **User-Agent** header, so the server is not overloaded.
- **Purpose**: the data is used **only for learning / this course assignment**.
- **Reuse**: existing results in `data/crawling_detik.csv` / `.json` are loaded instead of re-crawling; re-crawl stays disabled by default (`_re_crawl = False`).
