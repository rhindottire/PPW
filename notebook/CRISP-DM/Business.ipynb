# [Business Understanding](https://rhindottire.github.io/Data-Mining/dataProcess.html#business-understanding)

This page documents the business understanding for the assignment series: the problem, the plan, the stakeholders, the success criteria, and the limits that apply. It deliberately states only the plan — dataset statistics, model results, and other evidence are recorded later in the stages where they are produced (Data Understanding, Input, Modeling, Output, and Production).

## 1. Define the Problem

Indonesian news articles from detik.com fall into two categories, sport and finance, and the course asks whether this categorization can be learned from text alone. The end goal is twofold: a **deployed application that classifies a pasted news link**, and **evidence about which text representation works better** for this task under an identical protocol.

The problem is answered step by step across five assignments: build a labeled news dataset, preprocess the text transparently, establish a classification baseline, learn word embeddings, and finally compare the two representations behind one application.

## 2. Set Objectives

1. **News dataset** — collect a labeled detik.com dataset covering sport and finance with the required schema (`id`, `isi_berita`, `label`, `url`), following the crawling ethics below.
2. **Text preprocessing** — clean and transform the raw text with every step shown before and after it runs, so the pipeline is reproducible and open to inspection.
3. **Data classification** — build a classification baseline from a sparse TF-IDF representation and Gaussian Naive Bayes, the classifier required by the course.
4. **Word embedding** — learn a dense skip-gram representation of the same texts and pair it with the same classifier family.
5. **Comparison and application** — tune the skip-gram model with grid search, compare both representations under one shared protocol, export the chosen model and classifier as `.pkl` files, build an application that accepts a pasted news link, and deploy it on a Hugging Face Space.

The remaining CRISP-DM notebooks carry out the final comparison in order: `Input.ipynb` prepares the data, `Modeling.ipynb` runs the tuning, `Output.ipynb` evaluates the comparison, and `Production.ipynb` exports the models and deploys the application.

## 3. Identify Stakeholders

- **Student (me)**: carries out all five assignments and the deployment as deliverables of the practical course.
- **Lecturer (MULA'AB, S.Si., M.Kom)**: evaluates whether each assignment and the deployed application meet the requirements.
- **Readers** of the published Jupyter Book / GitHub Pages: need the plan, the comparison, and the ethics limits stated plainly so the work can be checked.
- **Hugging Face Space users**: paste a news link and expect a sport/finance prediction, while the fetched site keeps its robots.txt allowance and rate limits respected.

## 4. Define Success Criteria

- Every transformation is demonstrated with examples or counts before and after it runs; the pipeline can be repeated from the notebook.
- The final comparison is decided by **executed numbers**, not by assumption, and the evaluation states plainly which representation wins.
- **No test-set leakage**: the split happens before any fitting, and the embedding model is retrained inside every cross-validation fold.
- The exported `.pkl` files **reload and reproduce the notebook predictions** exactly.
- The application classifies a pasted URL with the **same preprocessing** used during training and is reachable on the Hugging Face Space.
- Live fetching stays within the ethics limits below: a handful of URLs, at least 1 second apart, honest client identification.

## 5. Assess Situation & Ethics

The assignments and the demonstration run under the following risks, constraints, and ethical limits.

### Risks and Constraints

- **Small evaluation set**: one accuracy figure can mislead; cross-validation is mandatory to stabilize the comparison.
- **Pipeline sensitivity**: preprocessing choices change measured accuracy, so both representations must be locked to one pipeline before the comparison is run.
- **Compute**: every grid-search fold retrains the embedding model, so tuning is CPU-bound and runs for minutes; no GPU is involved or needed.
- **Environment**: the local environment and the deployed Space use separate dependency sets; the Space pins its own small requirements file instead of reusing the course `requirements.txt`.
- **Connectivity**: the live demo and the deployment step require internet access during notebook execution.

### Ethics

- **robots.txt**: detik.com allows crawling (`User-agent: * Allow: /`); no page that robots disallows is fetched.
- **Volume and pace**: the bulk dataset is collected once with at least **1 second** between requests; the demonstration fetches only **2-3 individual article URLs** the same way.
- **Honest identification**: all requests send a **descriptive User-Agent** naming the project and contact URL — no browser impersonation, no header spoofing, no bypass of any protection.
- **Reuse**: the dataset in `data/Web-Mining/` is loaded instead of being crawled again; a re-crawl stays disabled by default.
- **Application behavior**: the deployed app fetches **one URL per user request, on demand**, and stores nothing beyond the extracted text needed for the prediction shown on screen.
- **Authorization status**: no separate written permission was requested from the site; access relies on the public robots.txt allowance and the non-commercial educational purpose stated here. This limitation is recorded rather than hidden.
- **Purpose**: all outputs — models, notebook, and application — are used **only for learning / this course assignment**.
