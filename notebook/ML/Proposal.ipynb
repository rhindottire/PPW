## 1. Business Understanding

Fase pertama metodologi CRISP-DM berfokus pada pemahaman mendalam terhadap masalah fundamental institusi pendidikan tinggi, perumusan sasaran bisnis dan teknis, pemetaan pemangku kepentingan (*stakeholders*), serta penetapan kriteria keberhasilan proyek.

### 1.1 Define the Problem

Fenomena putus studi (*student dropout*) merupakan tantangan struktural kritis yang dihadapi oleh institusi pendidikan tinggi di seluruh dunia. Kegagalan mahasiswa dalam menyelesaikan studi menimbulkan kerugian multidimensi yang signifikan:
1. **Bagi Mahasiswa**: Kerugian finansial pribadi, hilangnya waktu produktif, dan penurunan peluang mobilitas sosial-ekonomi di masa depan.
2. **Bagi Universitas**: Penurunan efisiensi operasional, penurunan skor akreditasi institusi, pemborosan kapasitas kuota kelas, serta hilangnya pendapatan biaya kuliah (SPP).
3. **Bagi Negara**: Pemborosan subsidi pendidikan tinggi publik dan terhambatnya target pembentukan modal manusia (*human capital*) berdaya saing global.

Pendekatan manajerial konvensional yang bersifat reaktif—seperti memanggil mahasiswa setelah nilai IPK anjlok di semester akhir atau setelah menunggak pembayaran SPP berbulan-bulan—terbukti tidak efektif karena intervensi datang terlambat. Oleh karena itu, diperlukan pendekatan prediktif menggunakan **Machine Learning** untuk mendeteksi indikasi dini risiko putus studi pada semester 1 dan semester 2, sehingga pihak kampus memiliki jendela waktu intervensi preventif yang memadai.

### 1.2 Set Objectives

Proyek Machine Learning ini menetapkan dua pilar sasaran yang saling mendukung:
1. **Sasaran Bisnis (Manajerial Kampus)**: Membangun sistem peringatan dini (*Early Warning System*) berbasis analitik prediktif untuk memetakan mahasiswa berisiko tinggi putus studi, sehingga pihak kampus dapat mengalokasikan program bimbingan akademik, restrukturisasi skema pembayaran SPP, dan bantuan beasiswa secara proaktif dan tepat sasaran.
2. **Sasaran Machine Learning**: Mengembangkan dan membandingkan model klasifikasi biner (*Binary Classification*) untuk memprediksi apakah seorang mahasiswa akan berhasil lulus (`Graduate: 0`) atau berisiko mengalami putus studi (`Dropout: 1`) berdasarkan indikator demografis, sosio-ekonomi, dan performa kurikuler awal.

**Justifikasi Pemilihan Klasifikasi Dibandingkan Regresi**:
Pemilihan pendekatan **Supervised Classification** dan bukan **Regression** didasari oleh tiga landasan teoritis dan praktis yang mendasar:
- **Sifat Alami Variabel Target (Diskrit Kategorikal)**: Kolom `Target` pada dataset berisi status administratif kualitatif: `Dropout`, `Enrolled`, dan `Graduate`. Ini adalah label kualitatif nominal, bukan besaran angka kontinu. Kita tidak bisa meregresikan kata 'Lulus' atau 'Dropout' karena tidak ada makna fisik dari nilai kontinu seperti *'mahasiswa ini lulus 0,73'* atau *'mahasiswa ini DO 1,45'*.
- **Kebutuhan Intervensi Manajerial Kampus (Actionable Decision Boundary)**: Pihak kampus dan dosen wali memerlukan batas keputusan (*decision boundary*) yang tegas untuk memicu *Standard Operating Procedure (SOP)* penanganan (seperti panggilan konseling akademik darurat atau restrukturisasi cicilan SPP bagi mahasiswa berlabel 'Dropout'). Model regresi yang sekadar memprediksi angka IPK akhir tidak secara langsung menjawab pertanyaan kritis apakah mahasiswa bersangkutan akan bertahan atau keluar.
- **Pelanggaran Asumsi Matematis Regresi Linear**: Menerapkan regresi pada target biner (*Linear Probability Model*) menghasilkan prediksi nilai yang melanggar batas probabilitas logis (bisa bernilai $< 0$ atau $> 1$) serta melanggar asumsi homoskedastisitas residual. Sebaliknya, model klasifikasi berbasis probabilitas posterior terkalibrasi memungkinkan sistem melakukan pemeringkatan risiko (*risk triage*) bertingkat secara presisi.

- **Mahasiswa**: Memperoleh pendampingan akademik dan dukungan finansial sebelum terlanjur terancam sanksi putus studi.
- **Tim Pengembang Proyek (Kelompok)**: Mahasiswa pelaksana yang bertanggung jawab mengeksekusi pipeline CRISP-DM dari eksplorasi data, rekayasa fitur, pemodelan, hingga pelaporan ilmiah.

### 1.4 Define Success Criteria

Keberhasilan implementasi proyek diukur berdasarkan kriteria kuantitatif dan kualitatif berikut:
1. **Kriteria Evaluasi Teknis Machine Learning**:
   - **Recall Kelas Dropout $\ge 85\%$**: Mengutamakan sensitivitas deteksi terhadap kelas minoritas mahasiswa rentan DO guna meminimalkan *False Negative* (mahasiswa berisiko yang luput dari pantauan sistem).
   - **F1-Score $\ge 88\%$**: Menjaga keseimbangan harmonis antara ketepatan deteksi (*Precision*) dan cakupan deteksi (*Recall*).
   - **Akurasi Model Uji $\ge 88\%$**: Mengungguli secara signifikan batas *The Accuracy Trap* (baseline tebakan kelas mayoritas sebesar 60,85%).
   - **ROC-AUC $\ge 0,90$**: Memastikan daya pemisah probabilitas model berada pada kategori *Excellent Discrimination*.
2. **Kriteria Transparansi & Operasional**:
   - Menghasilkan aturan pohon keputusan (*white-box decision rules*) yang konkret dan mudah dipahami oleh pengambil kebijakan non-teknis di lingkungan kampus.

**Penjelasan Parameter Institusional Riil Ketiga Label Target**:
Berdasarkan dokumentasi resmi institusi dan publikasi ilmiah aslinya, status kelulusan mahasiswa diekstraksi pada batas akhir durasi kurikulum normal sarjana (3 tahun / 6 semester) dengan tiga kategori institusional:
1. **`Graduate` (Lulus)**: Mahasiswa yang berhasil menyelesaikan seluruh beban kurikulum penuh (180 ECTS / SKS sistem Eropa) pada akhir masa studi dan secara administratif dinyatakan lulus serta berhak menyandang gelar sarjana.
2. **`Dropout` (Putus Studi / Keluar)**: Mahasiswa yang secara formal mengundurkan diri (*deregistered*) ATAU mahasiswa yang tidak melakukan registrasi ulang / menunggak pembayaran SPP (*tuition fees overdue*) melebihi batas semester toleransi sehingga hak akademiknya ditutup permanen oleh kampus sebelum mencapai 180 ECTS.
3. **`Enrolled` (Masih Terdaftar Aktif)**: Mahasiswa yang pada saat akhir durasi normal kurikulum status registrasinya masih tercatat aktif kuliah (karena mengulang mata kuliah yang belum lulus atau sedang menyelesaikan tugas akhir). Mahasiswa ini belum memenuhi 180 ECTS, namun belum mengundurkan diri dan tetap membayar biaya registrasi.

**Alasan Ilmiah Pemisahan Label `Enrolled`**:
Dalam perumusan **Binary Classification** (*Dropout vs Graduate*), kelas `Enrolled` merupakan status transisi yang sedang berjalan (*censored ongoing process*), bukan takdir akhir (*terminal outcome*). Memasukkan data `Enrolled` ke dalam data latih klasifikasi biner akan menimbulkan ambiguitas label (*label noise*) yang menurunkan akurasi model menjadi ~77%. Oleh karena itu, **794 baris data mahasiswa berstatus Enrolled disisihkan dari data latih/uji klasifikasi biner, dan dialokasikan khusus sebagai kolam simulasi inferensi sistem deteksi dini (*Early Warning System*) pada tahap deployment.**

**A. Target Distribution Analysis**

Memeriksa persebaran frekuensi kelas pada dataset mentah (3 kelas) dan perbandingannya dengan formulasi biner definitif:

```python
target_counts = df['Target'].value_counts()
target_pct = (df['Target'].value_counts(normalize=True) * 100).round(2)

target_summary = pd.DataFrame({
    'Target Class': target_counts.index,
    'Count (Students)': target_counts.values,
    'Percentage (%)': target_pct.values
})
target_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 4.5))
palette = {'Graduate': '#2ca02c', 'Dropout': '#d62728', 'Enrolled': '#1f77b4'}
ax = sns.countplot(
    data=df,
    x='Target',
    order=['Graduate', 'Dropout', 'Enrolled'],
    hue='Target',
    palette=palette,
    legend=False
)

plt.title('Distribution of Student Academic Outcomes', fontsize=12, pad=12)
plt.xlabel('Academic Outcome (Target)', fontsize=10)
plt.ylabel('Number of Students', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

for p in ax.patches:
    height = int(p.get_height())
    pct = (height / len(df)) * 100
    ax.annotate(f'{height:,}\n({pct:.1f}%)', 
                (p.get_x() + p.get_width() / 2., height / 2),
                ha='center', va='center', fontsize=10, color='white', weight='bold')

plt.tight_layout()
plt.show()
```

Diagram batang di atas memperlihatkan komposisi kelas mentah: **Graduate 49,9% (2.209 mahasiswa)**, **Dropout 32,1% (1.421 mahasiswa)**, dan **Enrolled 18,0% (794 mahasiswa)**. **B. Academic Correlation Matrix**

Memeriksa korelasi Pearson antar-variabel kurikuler pada semester 1 dan semester 2 untuk mendeteksi multikolinearitas:

**C. Financial Risk Analysis**

Menganalisis dampak tunggakan biaya kuliah (SPP) dan status debitur utang terhadap risiko putus studi:

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
palette = {'Graduate': '#2ca02c', 'Dropout': '#d62728', 'Enrolled': '#1f77b4'}

# 1. Tuition fees status
tuition_pct = pd.crosstab(df['Tuition fees up to date'], df['Target'], normalize='index') * 100
tuition_pct = tuition_pct[['Graduate', 'Dropout', 'Enrolled']]
tuition_pct.plot(
    kind='bar',
    ax=axes[0],
    color=[palette[c] for c in tuition_pct.columns],
    width=0.7,
    edgecolor='black',
    linewidth=0.5
)
axes[0].set_title('Academic Outcome by Tuition Fees Status', fontsize=11, pad=10)
axes[0].set_xlabel('Tuition Fees Status', fontsize=10)
axes[0].set_ylabel('Percentage within Group (%)', fontsize=10)
axes[0].set_xticklabels(['Unpaid / In Arrears (0)', 'Up to Date (1)'], rotation=0)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].legend(title='Outcome', loc='upper right')

for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.1f%%', label_type='edge', padding=2, fontsize=8.5)

# 2. Debtor status
debtor_pct = pd.crosstab(df['Debtor'], df['Target'], normalize='index') * 100
debtor_pct = debtor_pct[['Graduate', 'Dropout', 'Enrolled']]
debtor_pct.plot(
    kind='bar',
    ax=axes[1],
    color=[palette[c] for c in debtor_pct.columns],
    width=0.7,
    edgecolor='black',
    linewidth=0.5
)
axes[1].set_title('Academic Outcome by Debtor Status', fontsize=11, pad=10)
axes[1].set_xlabel('Debtor Status', fontsize=10)
axes[1].set_ylabel('Percentage within Group (%)', fontsize=10)
axes[1].set_xticklabels(['No Debt (0)', 'Has Debt (1)'], rotation=0)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].legend(title='Outcome', loc='upper right')

for container in axes[1].containers:
    axes[1].bar_label(container, fmt='%.1f%%', label_type='edge', padding=2, fontsize=8.5)

plt.tight_layout()
plt.show()
```

Grafik di atas membuktikan bahwa faktor finansial merupakan salah satu pemisah paling tajam:
- Mahasiswa yang menunggak pembayaran SPP (*Tuition fees = 0*) memiliki angka **Dropout melonjak hingga 86,6%** dengan tingkat kelulusan hanya 6,8%.
- Mahasiswa dengan status berutang (*Debtor = 1*) memiliki angka **Dropout sebesar 61,4%**, dua kali lipat lebih tinggi dari mahasiswa tanpa utang (28,3%).

Mengevaluasi peran protektif bantuan pembiayaan beasiswa terhadap stabilitas kelulusan mahasiswa:

```python
scholar_pct = pd.crosstab(df['Scholarship holder'], df['Target'], normalize='index') * 100
scholar_pct = scholar_pct[['Graduate', 'Dropout', 'Enrolled']]

ax = scholar_pct.plot(
    kind='bar',
    color=[palette[c] for c in scholar_pct.columns],
    width=0.6,
    edgecolor='black',
    linewidth=0.5,
    figsize=(7, 4.5)
)

plt.title('Academic Outcome Distribution by Scholarship Status', fontsize=12, pad=12)
plt.xlabel('Scholarship Status', fontsize=10)
plt.ylabel('Percentage within Group (%)', fontsize=10)
plt.xticks([0, 1], ['Non-Scholarship Holder (0)', 'Scholarship Holder (1)'], rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Outcome', loc='upper right')

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', label_type='edge', padding=2, fontsize=8.5)

plt.tight_layout()
plt.show()
```

**2. Outlier Analysis (Metode IQR) pada Usia Pendaftaran**

Mendeteksi pencilan statistik pada usia pendaftaran menggunakan metode *Interquartile Range* (IQR) dan mengevaluasi justifikasi ilmiahnya:

```python
# Analisis Pencilan Usia dengan Metode IQR
q1_age = df['Age at enrollment'].quantile(0.25)
q3_age = df['Age at enrollment'].quantile(0.75)
iqr_age = q3_age - q1_age
upper_bound_age = q3_age + 1.5 * iqr_age
outliers_age = df[df['Age at enrollment'] > upper_bound_age]

outlier_age_df = pd.DataFrame({
    'Parameter Statistik IQR': [
        'Kuartil 1 (Q1 - 25%)',
        'Median (Q2 - 50%)',
        'Kuartil 3 (Q3 - 75%)',
        'Interquartile Range (IQR)',
        'Ambang Batas Atas (Q3 + 1.5*IQR)',
        'Jumlah Outlier Statistik (Usia > 34 Thn)',
        'Persentase Outlier terhadap Populasi',
        'Usia Maksimum dalam Dataset'
    ],
    'Nilai': [
        f"{q1_age:.1f} tahun",
        f"{df['Age at enrollment'].median():.1f} tahun",
        f"{q3_age:.1f} tahun",
        f"{iqr_age:.1f} tahun",
        f"{upper_bound_age:.1f} tahun",
        f"{len(outliers_age):,} mahasiswa",
        f"{len(outliers_age) / len(df) * 100:.2f}%",
        f"{df['Age at enrollment'].max()} tahun"
    ]
})
outlier_age_df
```

**Justifikasi Domain terhadap Outlier Usia (*No Trimming Policy*)**:
Secara matematis murni, rumus IQR menetapkan batas atas usia sebesar **34,0 tahun**, sehingga sebanyak **442 mahasiswa (9,99%)** terdeteksi sebagai *outlier statistik*. Namun, dalam konteks ilmu data terapan (*applied data science*):
1. **Bukan Kesalahan Pengukuran**: Nilai usia 35 hingga 70 tahun adalah data manusia nyata yang mendaftar secara sah melalui program jalur dewasa (*Maiores de 23*).
2. **Bahaya Pemotongan (*Trimming Danger*)**: Menghapus 442 mahasiswa ini akan membuang 10% data riil dan menciptakan *blind spot* (kebutaan model) di mana sistem tidak akan mampu mengenali risiko mahasiswa dewasa di dunia nyata.
3. **Keputusan Preprocessing**: Seluruh data usia dipertahankan 100% tanpa pemotongan. Untuk algoritma Decision Tree, skala usia tidak mempengaruhi pemisahan (*split*); sedangkan untuk algoritma KNN, fitur usia akan distandarisasi (*StandardScaler*) pada Data Preparation agar besaran angka tidak mendistorsi perhitungan jarak Euclidean.

**3. Class Imbalance Analysis & The Accuracy Trap**

Menganalisis proporsi kelas biner definitif dan membongkar jebakan akurasi semu (*The Accuracy Trap*):

```python
# Analisis The Accuracy Trap pada Populasi Biner Definitif
df_bin = df[df['Target'].isin(['Graduate', 'Dropout'])]
n_grad = sum(df_bin['Target'] == 'Graduate')
n_drop = sum(df_bin['Target'] == 'Dropout')
total_bin = len(df_bin)

dummy_accuracy = (n_grad / total_bin) * 100
imbalance_ratio = n_grad / n_drop

imbalance_df = pd.DataFrame({
    'Metrik Kelas Biner': [
        'Total Sampel Definitif',
        'Kelas Mayoritas: Graduate (0)',
        'Kelas Minoritas: Dropout (1)',
        'Rasio Ketidakseimbangan (Disproportion)',
        'The Accuracy Trap (Dummy Classifier Baseline)',
        'Konsekuensi Evaluasi'
    ],
    'Nilai Kuantitatif': [
        f"{total_bin:,} mahasiswa",
        f"{n_grad:,} ({n_grad / total_bin * 100:.2f}%)",
        f"{n_drop:,} ({n_drop / total_bin * 100:.2f}%)",
        f"{imbalance_ratio:.2f} : 1 (~61% vs ~39%)",
        f"{dummy_accuracy:.2f}% (Tebak Graduate Selalu)",
        "Akurasi mentah menyesatkan; Wajib Recall & F1-Score"
    ]
})
imbalance_df
```

**Analisis Dampak Ketidakseimbangan Kelas (*The Accuracy Trap*)**:
Pada populasi 3.630 data biner definitif, proporsi kelas adalah **60,85% Graduate** dan **39,15% Dropout** (rasio 1,55 : 1):
1. **Bukan Extreme Imbalance**: Rasio 60:40 tergolong *mild imbalance* yang merupakan cerminan sehat dunia pendidikan (lebih banyak yang lulus daripada DO). Oleh karena itu, teknik manipulasi sintetis agresif seperti **SMOTE tidak direkomendasikan** karena berisiko menginterpolasi data palsu pada variabel biner (seperti beasiswa dan status debitur).
2. **Bahaya The Accuracy Trap**: Jika model bersikap malas dan selalu memprediksi 'Graduate' untuk semua mahasiswa (*Zero-Rule / Dummy Classifier*), model akan mendapatkan **akurasi semu sebesar 60,85%** padahal gagal total mendeteksi 1 orang pun mahasiswa yang akan DO (*Recall = 0%*).
3. **Solusi Metodologis**: Menetapkan pembobotan biaya (`class_weight='balanced'`) pada Decision Tree dan memprioritaskan metrik **Recall** dan **F1-Score** pada kelas Dropout agar model dihukum lebih berat jika melewatkan mahasiswa yang berisiko.
