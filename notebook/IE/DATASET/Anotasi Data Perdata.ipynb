```python
def clean_text(text):
  text = text.replace("\n", " ")
  text = text.replace("P U T U S A N", "PUTUSAN").replace("P E N E T A P A N", "PENETAPAN")
  text = text.replace('\uf0d8', '')
  text = re.sub(r'\b(\w)\s*\b', r'\1', text)
  text = re.sub(r'(Hal\.\s*\S+(?:\s*\S+)?\.?\s*)|(Halaman \d+(?:\.\d+)?\s*)|Putusan Nomor \S+\s*', '', text)
  text = re.sub(r'^[A-Za-z]+\.\s*', '', text)
  text = re.sub(r'-+', '', text)
  text = re.sub(r'\s+', ' ', text)
  text = re.sub(r'(pn)\.', r'\1 ', text) # Memperbaiki nomor putusan mengganti titik setelah PN dengan spasi
  text = re.sub(r'(\d+)\s*([a-zA-Z]+)\s*(\d+)', r'\1 \2 \3', text) # Memperbaiki format tanggal jika 1 angka di kasih 0
  text = re.sub(r'(\b\d\b)', lambda x: f'0{x.group(1)}', text)
  text = re.sub(r"(\d+)\.(\w)", r"\1. \2", text)
  text = re.sub(r'\b(jkt) (pst)\b', r'\1.\2', text)
  return text.lower()

df_200['content'] = df_200["content"].apply(clean_text)
```

```python
def find_similar_sentence(text, target_sentence):
  sentences = text.split(';')  # Pisahkan teks menjadi kalimat-kalimat
  if len(target_sentence.split()) > 1:
    # Gunakan TF-IDF untuk mewakili teks
    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform(sentences + [target_sentence])

    # Hitung cosine similarity antara vektor target dan vektor kalimat
    similarity_scores = cosine_similarity(vectors[-1], vectors[:-1])[0]

    # Temukan kalimat dengan similarity tertinggi
    max_similarity_index = similarity_scores.argmax()

    if similarity_scores[max_similarity_index] > 0.0:  # Ubah ambang batas sesuai kebutuhan
        return sentences[max_similarity_index].strip()
    else:
        return None
  else:
    return target_sentence
```

```python
def generate_label(data):
  label = {}
  nomor_putusan = data["nomor"].lower()
  tgl_putusan = re.sub(r'(\b\d\b)', lambda x: f'0{x.group(1)}', data["tanggal_dibacakan"].lower())
  hakim_ketua = data["hakim_ketua"].lower().replace("hakim ketua", '').strip()
  hakim_anggota = data["hakim_anggota"].lower().replace("hakim anggota", '').replace("br", '').strip().split(',')
  panitera = data["panitera"].lower().replace("panitera pengganti", "").replace(":", "").strip()
  penggugat = re.findall(r'\d+\.(.*?)\s*(?=\d+\.|$)', data["penggugat"].lower())
  tergugat = re.findall(r'\d+\.(.*?)\s*(?=\d+\.|$)', data["tergugat"].lower())
  turut_tergugat = re.findall(r'\d+\.(.*?)\s*(?=\d+\.|$)', data["turut_tergugat"].lower()) if str(data["turut_tergugat"]) != 'nan' else None
  amar_lainnya = data["amar_lainnya"].lower()
  catatan_amar = re.sub(r'(\w+)\s*(\d+\.)', r'\1 \2', re.sub(r'(pn)\.', r'\1 ', data["catatan_amar"].lower()))
  hasil_gugatan = amar_lainnya if re.search(amar_lainnya, data["content"]) else re.sub(r'\d+\.', '', find_similar_sentence(catatan_amar, amar_lainnya)).strip()
  putusan_gugatan = [i.strip() for i in re.sub(r'(pn)\.', r'\1 ', data["catatan_amar"].lower()).split(';') if re.search('menyatakan|dalam pokok perkara|menghukum para|menghukum kedua', i.strip())]
  # print(label[nomor_putusan])
  label[nomor_putusan] = "B_VERN"
  label[tgl_putusan] = "B_TIMV"
  label[hakim_ketua] = "B_JUDP"
  label[panitera] = "B_REGI"
  label[hasil_gugatan] = "B_LSWR"

  if putusan_gugatan:
    if len(putusan_gugatan) > 1:
      label.update({pg.strip() : "B_LSWV" for pg in putusan_gugatan})
    else:
      label[putusan_gugatan[0]] = "B_LSWV"

  if not hakim_anggota:
    label[data["hakim_anggota"].lower()] = "B_PLAN"
  else:
    label.update({anggota.strip() : "B_JUG" for anggota in hakim_anggota})

  if not penggugat:
    label[data["penggugat"].lower()] = "B_PLAN"
  else:
    label.update({i.strip() : "B_PLAN" for i in penggugat})

  if not tergugat:
    label[data["tergugat"].lower()] = "B_DENN"
  else:
    label.update({i.strip() : "B_DENN" for i in tergugat if i})

  if turut_tergugat != None and not turut_tergugat:
    label[data["turut_tergugat"].lower() if re.search(data["turut_tergugat"].lower(), data["content"]) else re.sub(r'\d+\.', '', find_similar_sentence(data["content"], data["turut_tergugat"].lower())).strip()] = "B_PLAO"
  if turut_tergugat:
    label.update({i.strip() : "B_PLAO" for i in turut_tergugat if i})

  return label

def labeling_token(text, label_dict):
  tokens = text.split()
  labeled_tokens = ['O'] * len(tokens)
  find_label = dict()
  for key, value in label_dict.items():
      words = key.split()
      # print(words, value)
      current_dict = find_label
      for word in words[:-1]:
          word = re.sub(r'[.,;]', '', word)
          current_dict = current_dict.setdefault(word, {})
      last_word = re.sub(r'[.,;]', '', words[-1])
      current_dict[last_word] = {'label': value}
  # print(find_label)

  temp_idx_token = []
  find_label_selected = find_label
  for i, token in enumerate(tokens):
    clean_token = re.sub(r'[.,;]', '', token)
    # print(clean_token)
    if clean_token in find_label_selected:
      find_label_selected = find_label_selected[clean_token]
      temp_idx_token.append(i)
      # print(clean_token)
      # print(find_label_selected)
      if 'label' in find_label_selected:
        tag = find_label_selected['label']
        # print(clean_token, tag, temp_idx_token)
        new_label = [tag if ind == 0 else tag.replace('B', 'I') for ind in range(len(temp_idx_token))]
        labeled_tokens[temp_idx_token[0]:temp_idx_token[0] + len(new_label)] = new_label
        temp_idx_token = []
        find_label_selected = find_label
    else:
      temp_idx_token = []
      find_label_selected = find_label

  return labeled_tokens
def countLabel(result, label_ori):
  label_found = {}
  temp = []
  for i, row in result.iterrows():
    for t, l in zip(row["text"], row["label"]):
      if l != 'O' and l.replace('I_', 'B_') not in temp:
        temp.append(l.replace('I_', 'B_'))

  label_found[f"Dokumen Ke-{i+1}"] = (len(temp), label_ori, temp)

  return label_found

def accuracy(result):
  sum_label = 0
  label_in = 0
  for key, value in result.items():
    print(f"{key} => Data Terlebel : {value[0]} ; Total Label : {value[1]} ; Label : {value[2]}")
    sum_label+= value[1]
    label_in += value[0]
  print("Accuracy : {label_in/sum_label * 100}")

  return f"Accuracy : {label_in/sum_label * 100}"

def anotasi(data):
  new_data = pd.DataFrame([], columns=['doc','text', 'label'])
  start = 0
  end = 0
  count_label = {}
  for i, row in tqdm(data.iterrows(), total=len(data), desc="Annotating"):
    label = generate_label(row)

    labeled_tokens = labeling_token(row["content"], label)
    new_data = pd.concat([new_data, pd.DataFrame({"doc": [row["nama_file"].split()],"text": [row["content"].split()], "label": [labeled_tokens]})], ignore_index=True)
    new_data = new_data.explode('doc').reset_index(drop=True)
    count_label.update(countLabel(new_data.loc[i:i], len(set(label.values()))))

  return (new_data, count_label)


angka = 47
angka1 = 47
angka2 = 100
new_df, count_label = anotasi(df_200)
df.loc[angka1:angka2].apply(generate_label, axis=1)[angka]
# generate_label(df_1)
```
