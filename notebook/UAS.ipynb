# Final Examination

Dokumen ini dialokasikan sebagai modul pelaporan untuk Evaluasi Akhir Semester (UAS) pada mata kuliah Pencarian & Penambangan Web (PPW).

## Examination Scope

UAS dirancang untuk menguji penguasaan komprehensif terhadap seluruh siklus penambangan web, mencakup:

- Ekstraksi dan pengumpulan data web berskala penuh sesuai etika crawling.
- Pipeline pembersihan dan rekayasa fitur teks tingkat lanjut (vektorisasi dan word embedding).
- Evaluasi pemodelan klasifikasi, pengelompokan (clustering), atau penataan informasi lanjutan.
- Integrasi dan publikasi sistem inferensi cerdas berbasis web.

## Project Status

Saat ini perkuliahan masih berlangsung pada materi reguler sebelum pekan evaluasi akhir semester. Seluruh deliverable praktikum reguler (Book 1 hingga Book 5) serta proyek UTS telah selesai dikerjakan dan terdokumentasi lengkap pada bab-bab sebelumnya.

Implementasi kode dan analisis mendalam untuk proyek akhir akan dipublikasikan pada halaman ini begitu instruksi spesifik UAS dirilis secara resmi oleh dosen pengampu.

## Conclusion

Halaman ini bertindak sebagai penanda resmi kesiapan infrastruktur web book untuk menerima artefak akhir semester. Seluruh lingkungan runtime (`enWebmining`), dependensi model, dan repositori telah tervalidasi siap untuk pengerjaan UAS mendatang.
