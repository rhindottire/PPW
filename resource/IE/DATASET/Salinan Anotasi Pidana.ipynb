```python
def clean_text(text):
  text = text.replace("\n", " ")
  text = text.replace("P U T U S A N", "PUTUSAN").replace("P E N E T A P A N", "PENETAPAN")
  text = text.replace('\uf0d8', '')
  text = re.sub(r'\b(\w)\s*\b', r'\1', text)
  text = re.sub(r'(Hal\.\s*\S+(?:\s*\S+)?\.?\s*)|(Halaman \d+(?:\.\d+)?\s*)|Putusan Nomor \S+\s*', '', text)
  text = re.sub(r'^[A-Za-z]+\.\s*', '', text)
  text = re.sub(r'-+', '', text)
  text = re.sub(r'\s+', ' ', text)
  text = re.sub(r'(pn)\.', r'\1 ', text) # Memperbaiki nomor putusan mengganti titik setelah PN dengan spasi
  text = re.sub(r'(\d+)\s*([a-zA-Z]+)\s*(\d+)', r'\1 \2 \3', text) # Memperbaiki format tanggal jika 1 angka di kasih 0
  text = re.sub(r'(\b\d\b)', lambda x: f'0{x.group(1)}', text)
  text = re.sub(r"(\d+)\.(\w)", r"\1. \2", text)
  text = re.sub(r'\b(jkt) (pst)\b', r'\1.\2', text)
  return text

df_200['content'] = df_200["content"].apply(clean_text)
```

```python
def find_similar_sentence(text, target_sentence):
  sentences = text.split(';')  # Pisahkan teks menjadi kalimat-kalimat
  if len(target_sentence.split()) > 1:
    # Gunakan TF-IDF untuk mewakili teks
    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform(sentences + [target_sentence])

    # Hitung cosine similarity antara vektor target dan vektor kalimat
    similarity_scores = cosine_similarity(vectors[-1], vectors[:-1])[0]

    # Temukan kalimat dengan similarity tertinggi
    max_similarity_index = similarity_scores.argmax()

    if similarity_scores[max_similarity_index] > 0.0:  # Ubah ambang batas sesuai kebutuhan
        return sentences[max_similarity_index].strip()
    else:
        return None
  else:
    return target_sentence
```
