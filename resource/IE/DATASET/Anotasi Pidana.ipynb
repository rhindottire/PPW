### Cleaning

```python
def clean_text(text):
  text = text.replace("\n", " ")
  text = text.replace("P U T U S A N", "PUTUSAN").replace("P E N E T A P A N", "PENETAPAN")
  text = text.replace('\uf0d8', '')
  text = re.sub(r'\b(\w)\s*\b', r'\1', text)
  text = re.sub(r'(Hal\.\s*\S+(?:\s*\S+)?\.?\s*)|(Halaman \d+(?:\.\d+)?\s*)|Putusan Nomor \S+\s*', '', text)
  text = re.sub(r'^[A-Za-z]+\.\s*', '', text)
  text = re.sub(r'-+', '', text)
  text = re.sub(r'\s+', ' ', text)
  text = re.sub(r'(pn)\.', r'\1 ', text) # Memperbaiki nomor putusan mengganti titik setelah PN dengan spasi
  text = re.sub(r'(\d+)\s*([a-zA-Z]+)\s*(\d+)', r'\1 \2 \3', text) # Memperbaiki format tanggal jika 1 angka di kasih 0
  text = re.sub(r'(\b\d\b)', lambda x: f'0{x.group(1)}', text)
  text = re.sub(r"(\d+)\.(\w)", r"\1. \2", text)
  text = re.sub(r'\b(jkt) (pst)\b', r'\1.\2', text)
  text = re.sub(r'halaman\s*\d+\s*dari\s*\d+\s*', '', text) # Menghapus kata "halaman 1 dari ..."
  # text = re.sub(r'putusannomor(\d+/pid\.b/\d+/pnspg)', r'', text) # Menghapus bagian putusannomorXX/pid.b/YYYY/pnspg

  return text.lower()

df['content'] = df["content"].apply(clean_text)
```

### Anotasi Data

```python
def generate_label(data):
  label = {}
  nomor_putusan = data["nomor"].lower() # cek nomor putusan
  # cek terdakwa
  if any(char.isdigit() for char in data["terdakwa"]): # jika terdakwa lebih dari satu
    terdakwa = re.findall(r'\d+\.(.*?)\s*(?=\d+\.|$)', data["terdakwa"].lower().replace("(alm)", ''))
  else: # jika terdakwa hanya satu
    terdakwa = re.findall(r'(.*?)\s*(?=\d+\.|$)', data["terdakwa"].lower().replace("(alm)", ''))
  terdakwa_filtered = [item for item in terdakwa if item.strip()] # menghapus index kosong
  jenis_dakwaan = re.findall(r'tindak pidana\s+“([^“”]*)', data["content"].lower()) #cek dakwaan
  pasal = re.findall(r'pasal\s+([^;]*kuhp\b)', data["content"].lower())# cek pasal
  tgl_putusan = re.sub(r'(\b\d\b)', lambda x: f'0{x.group(1)}', data["tanggal_dibacakan"].lower()) # cek tanggal
  hakim_ketua = data["hakim_ketua"].lower().replace("hakim ketua", '').strip() # cek hakim ketua
  hakim_anggota = data["hakim_anggota"].lower().replace("hakim anggota", '').strip().split(',') # cek hakim anggota
  panitera = data["panitera"].lower().replace("panitera pengganti", "").replace(":", "").strip() # cek panitera
  penuntut_umum = re.findall(r'\d+\.(.*?)\s*(?=\d+\.|$)', data["penuntut_umum"].lower()) # cek penuntut umum
  amar_lainnya = data["amar_lainnya"].lower() # cek amar lainnya
  catatan_amar = re.sub(r'(\w+)\s*(\d+\.)', r'\1 \2', re.sub(r'(pn)\.', r'\1 ', data["catatan_amar"].lower().replace("mengadili:", ''))) # cek catatan amar
  tuntutan_hukuman = re.search(r'selama(.*?);', data["content"].lower()) # cek tuntutan hukuman
  if tuntutan_hukuman:
      tuntutan_hukuman = tuntutan_hukuman.group(1).strip() # menghilangkan re Match
  else:
      tuntutan_hukuman = ""
  putusan_hukuman = re.search(r'selama(.*?);', data["catatan_amar"].lower()) # cek putusan hukuman
  if putusan_hukuman:
      putusan_hukuman = putusan_hukuman.group(1).strip() # menghilangkan re Match
  else:
      putusan_hukuman = ""
  label[nomor_putusan] = "B_VERN"
  label[tgl_putusan] = "B_TIMV"
  label[hakim_ketua] = "B_JUDP"
  label[panitera] = "B_REGI"
  label[tuntutan_hukuman] = "B_PENA"
  label[putusan_hukuman] = "B_PUNI"
  label[pasal[0] if pasal else ""] = "B_ARTV"
  if not hakim_anggota:
    label[data["hakim_anggota"].lower()] = "B_JUG"
  else:
    label.update({anggota.strip() : "B_JUG" for anggota in hakim_anggota})
  if not penuntut_umum:
    label[data["penuntut_umum"].lower()] = "B_PROS"
  else:
    label.update({i.strip() : "B_PROS" for i in penuntut_umum})
  if not terdakwa_filtered:
    label[data["terdakwa"].lower()] = "B_DEFN"
  else:
    label.update({i.strip() : "B_DEFN" for i in terdakwa_filtered if i})
  if not jenis_dakwaan:
    label[data["kata_kunci"].lower()] = "B_CRIA"
  else:
    label.update({i.strip() : "B_CRIA" for i in jenis_dakwaan if i})
  return label

def labeling_token(text, label_dict):
  tokens = text.split()
  labeled_tokens = ['O'] * len(tokens)
  find_label = dict()
  for key, value in label_dict.items():
      words = key.split()
      # print(words, value)
      current_dict = find_label
      for word in words[:-1]:
          word = re.sub(r'[.,;]', '', word)
          current_dict = current_dict.setdefault(word, {})
      if words:
          last_word = re.sub(r'[.,;]', '', words[-1])
          current_dict[last_word] = {'label': value}
  # print(find_label)

  temp_idx_token = []
  find_label_selected = find_label
  for i, token in enumerate(tokens):
    clean_token = re.sub(r'[.,;]', '', token)
    # print(clean_token)
    if clean_token in find_label_selected:
      find_label_selected = find_label_selected[clean_token]
      temp_idx_token.append(i)
      # print(clean_token)
      # print(find_label_selected)
      if 'label' in find_label_selected:
        tag = find_label_selected['label']
        # print(clean_token, tag, temp_idx_token)
        new_label = [tag if ind == 0 else tag.replace('B', 'I') for ind in range(len(temp_idx_token))]
        labeled_tokens[temp_idx_token[0]:temp_idx_token[0] + len(new_label)] = new_label
        temp_idx_token = []
        find_label_selected = find_label
    else:
      temp_idx_token = []
      find_label_selected = find_label

  return labeled_tokens

def countLabel(result, label_ori):
  label_found = {}
  temp = []
  for i, row in result.iterrows():
    for t, l in zip(row["text"], row["label"]):
      if l != 'O' and l.replace('I_', 'B_') not in temp:
        temp.append(l.replace('I_', 'B_'))

  label_found[f"Dokumen Ke-{i+1}"] = (len(temp), label_ori, temp)

  return label_found

def accuracy(result):
  sum_label = 0
  label_in = 0
  for key, value in result.items():
    print(f"{key} => Data Terlebel : {value[0]} ; Total Label : {value[1]} ; Label : {value[2]}")
    sum_label+= value[1]
    label_in += value[0]
  print("Accuracy : {label_in/sum_label * 100}")

  return f"Accuracy : {label_in/sum_label * 100}"

def anotasi(data):
  new_data = pd.DataFrame([], columns=['doc','text', 'label'])
  # start = 0
  # end = 0
  count_label = {}
  for i, row in tqdm(data.iterrows(), total=len(data), desc="Annotating"):
    label = generate_label(row)
    labeled_tokens = labeling_token(row["content"], label)
    new_data = pd.concat([new_data, pd.DataFrame({"doc": [row["nama_file"].split()],"text": [row["content"].split()], "label": [labeled_tokens]})], ignore_index=True)
    new_data = new_data.explode('doc').reset_index(drop=True)
    count_label.update(countLabel(new_data.loc[i:i], len(set(label.values()))))

  return (new_data, count_label)
```
