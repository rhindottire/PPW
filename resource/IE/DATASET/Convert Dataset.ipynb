```python
# Pecah setiap teks dan label menjadi list berdasarkan koma dan spasi
split_text = [text.split(', ') for text in new_df['text']]
split_text_tag = [tag.split(', ') for tag in new_df['labels']]

# Periksa panjang setiap baris, pastikan jumlah elemen sama untuk kedua kolom
max_length = max(len(text) for text in split_text)
for i in range(len(split_text)):
    while len(split_text[i]) < max_length:
        split_text[i].append('')
    while len(split_text_tag[i]) < max_length:
        split_text_tag[i].append('')

# Buat DataFrame baru dengan kata dan label dalam baris terpisah
new_data2 = {'text': [word for sublist in split_text for word in sublist],
            'text-tag': [tag for sublist in split_text_tag for tag in sublist]}
new_df2 = pd.DataFrame(new_data2)

# Menampilkan DataFrame baru dengan kata dan label dalam baris terpisah
new_df2
```
