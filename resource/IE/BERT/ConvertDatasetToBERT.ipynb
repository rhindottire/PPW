```python
import pandas as pd
import csv
```

```python
df = pd.read_csv('/content/drive/MyDrive/IE-ML/NEW-DATASET/LEGALNER-POS-PREV-NEXT-200.csv')
df.head()
```

```python
wordToSentence=""
tagToSentence=""

f = open('/content/drive/MyDrive/BERT/LEGALBERTNER-POS-PREV-NEXT-200.csv','w')
w = csv.writer(f)
w.writerow(('text','labels'))

for index, row in df.iterrows():
    indicatorSentence = str(row["sentence"])

    if indicatorSentence!='nan' and index!=0:
        print(wordToSentence)
        print(tagToSentence)
        w.writerow([wordToSentence, tagToSentence])

        wordToSentence=""
        tagToSentence=""

    wordToSentence  = wordToSentence + str(row["word"]) + " "
    tagToSentence = tagToSentence + str(row["tag"]) + " "

print(wordToSentence)
print(tagToSentence)
w.writerow([wordToSentence,tagToSentence])

f.close()
```
